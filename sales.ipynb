```python

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 200]

products = ["Pen", "Book", "Pencil"]
values = [300, 500, 200]

fig, ax = plt.subplots(1, 3, figsize=(15, 4))

ax[0].plot(months, sales, marker="o")
ax[0].set_title("Monthly Sales")

ax[1].bar(products, values)
ax[1].set_title("Product Comparison")

ax[2].pie(values, labels=products, autopct="%1.1f%%")
ax[2].set_title("Sales Percentage")

plt.tight_layout()
plt.show()
```

```python

import numpy as np
import pandas as pd

data = {
    "Product": ["Pen", "Book", "Pen", "Pencil"],
    "Quantity": [10, 5, 20, 15],
    "Price": [10, 50, 10, 5]
}

df = pd.DataFrame(data)
df["Sales"] = df["Quantity"] * df["Price"]

print(df)
print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Maximum Sales:", df["Sales"].max())
print("Minimum Sales:", df["Sales"].min())
print("Standard Deviation:", np.std(df["Sales"]))

print(df.groupby("Product")["Sales"].sum())
```

```python

import tkinter as tk
from tkinter import messagebox


class StudentApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Student Management")
        self.students = []

        tk.Label(root, text="Student Name").pack()
        self.name = tk.Entry(root)
        self.name.pack()

        tk.Label(root, text="Student ID").pack()
        self.sid = tk.Entry(root)
        self.sid.pack()

        tk.Button(root, text="Add Student",
                  command=self.add_student).pack()

        tk.Button(root, text="Display Students",
                  command=self.display_students).pack()

        self.output = tk.Label(root, text="")
        self.output.pack()

    def add_student(self):
        name = self.name.get()
        sid = self.sid.get()

        if name and sid:
            self.students.append({"id": sid, "name": name})
            messagebox.showinfo("Success", "Student added")
            self.name.delete(0, tk.END)
            self.sid.delete(0, tk.END)
        else:
            messagebox.showwarning("Error", "Fill all fields")

    def display_students(self):
        text = ""
        for s in self.students:
            text += s["id"] + " - " + s["name"] + "\n"

        self.output.config(text=text or "No students")


root = tk.Tk()
app = StudentApp(root)
root.mainloop()
```

```python
# Simple Bank Account Program (OOP)

# Parent class
class Account:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance = self.balance - amount
            print("Withdrawn:", amount)
        else:
            print("Not enough balance")

    def check_balance(self):
        print(self.name, "- Balance:", self.balance)


# Child class 1 (inheritance): gets deposit, withdraw, check_balance from Account
class SavingsAccount(Account):
    # same method name, different rule (polymorphism)
    def withdraw(self, amount):
        if self.balance - amount >= 500:      # must keep Rs.500
            self.balance = self.balance - amount
            print("Withdrawn:", amount)
        else:
            print("Savings account must keep minimum Rs.500")


# Child class 2 (inheritance)
class CurrentAccount(Account):
    # same method name, different rule (polymorphism)
    def withdraw(self, amount):
        if amount <= self.balance + 5000:     # overdraft up to Rs.5000
            self.balance = self.balance - amount
            print("Withdrawn:", amount)
        else:
            print("Overdraft limit exceeded")


# Create two accounts
s = SavingsAccount("Ravi", 2000)
c = CurrentAccount("Priya", 1000)

# Menu
while True:
    print("\n1.Deposit  2.Withdraw  3.Balance  4.Exit")
    choice = input("Enter choice: ")

    if choice == "4":
        print("Thank you!")
        break

    print("Which account? 1.Savings  2.Current")
    t = input("Enter 1 or 2: ")
    if t == "1":
        acc = s
    else:
        acc = c

    if choice == "1":
        amt = int(input("Amount: "))
        acc.deposit(amt)
    elif choice == "2":
        amt = int(input("Amount: "))
        acc.withdraw(amt)      # same line, different result for Savings / Current
    elif choice == "3":
        acc.check_balance()
    else:
        print("Invalid choice")
```

```python

employees = []

while True:
    print("\n1.Add 2.Search 3.Display 4.Exit")
    choice = input("Enter choice: ")

    if choice == "1":
        emp_id = input("Enter ID: ")
        name = input("Enter name: ")
        salary = float(input("Enter salary: "))

        employees.append({
            "id": emp_id,
            "name": name,
            "salary": salary
        })
        print("Employee added")

    elif choice == "2":
        emp_id = input("Enter ID to search: ")
        found = False

        for emp in employees:
            if emp["id"] == emp_id:
                print(emp)
                found = True
                break

        if not found:
            print("Employee not found")

    elif choice == "3":
        for emp in employees:
            print(emp)

    elif choice == "4":
        break

    else:
        print("Invalid choice")
```

```python

menu = {"dosa": 40, "idli": 20, "tea": 10}
total = 0

print(menu)

while True:
    item = input("Enter item (done to finish): ").lower()

    if item == "done":
        break

    if item in menu:
        qty = int(input("Enter quantity: "))
        if qty > 0:
            total += menu[item] * qty
        else:
            print("Invalid quantity")
    else:
        print("Item not available")

if total > 500:
    discount = total * 0.10
elif total > 200:
    discount = total * 0.05
else:
    discount = 0

print("Total:", total)
print("Discount:", discount)
print("Final Bill:", total - discount)

```

```python

seats = 10
bookings = {}

while True:
    print("\n1.Book 2.Cancel 3.Check Seats 4.Display 5.Exit")
    choice = input("Enter choice: ")

    if choice == "1":
        if seats > 0:
            name = input("Enter name: ")
            bid = len(bookings) + 1
            bookings[bid] = name
            seats -= 1
            print("Ticket booked. ID:", bid)
        else:
            print("No seats available")

    elif choice == "2":
        bid = int(input("Enter booking ID: "))
        if bid in bookings:
            del bookings[bid]
            seats += 1
            print("Ticket cancelled")
        else:
            print("Invalid ID")

    elif choice == "3":
        print("Available seats:", seats)

    elif choice == "4":
        print(bookings)

    elif choice == "5":
        break

    else:
        print("Invalid choice")
```
