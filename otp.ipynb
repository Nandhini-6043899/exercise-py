```python

import random

otp = random.randint(100000, 999999)
print("Demo OTP:", otp)

for i in range(3):
    entered = input("Enter OTP: ")

    if entered == str(otp):
        print("Access Granted")
        break
    else:
        print("Wrong OTP")
else:
    print("Access Denied")
```

```python

menu = {"dosa": 40, "idli": 20, "tea": 10}
total = 0

print(menu)

while True:
    item = input("Enter item (done to finish): ").lower()

    if item == "done":
        break

    if item in menu:
        qty = int(input("Enter quantity: "))
        if qty > 0:
            total += menu[item] * qty
        else:
            print("Invalid quantity")
    else:
        print("Item not available")

if total > 500:
    discount = total * 0.10
elif total > 200:
    discount = total * 0.05
else:
    discount = 0

print("Total:", total)
print("Discount:", discount)
print("Final Bill:", total - discount)

```

```python

seats = 10
bookings = {}

while True:
    print("\n1.Book 2.Cancel 3.Check Seats 4.Display 5.Exit")
    choice = input("Enter choice: ")

    if choice == "1":
        if seats > 0:
            name = input("Enter name: ")
            bid = len(bookings) + 1
            bookings[bid] = name
            seats -= 1
            print("Ticket booked. ID:", bid)
        else:
            print("No seats available")

    elif choice == "2":
        bid = int(input("Enter booking ID: "))
        if bid in bookings:
            del bookings[bid]
            seats += 1
            print("Ticket cancelled")
        else:
            print("Invalid ID")

    elif choice == "3":
        print("Available seats:", seats)

    elif choice == "4":
        print(bookings)

    elif choice == "5":
        break

    else:
        print("Invalid choice")
```
